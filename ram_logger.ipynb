```python
import psutil
import time
import pandas as pd
from datetime import datetime
import os

INTERVAL = 5  # Collect RAM data every 5 seconds

file_path = "datas/ram_log.csv"

# Create the datas folder if it doesn't exist
os.makedirs("datas", exist_ok=True)

# Create CSV with column names if it doesn't already exist
if not os.path.exists(file_path):
    pd.DataFrame(columns=[
        "timestamp",
        "ram_used_mb",
        "ram_available_mb",
        "browser_ram_mb"
    ]).to_csv(file_path, index=False)

print("RAM logging started...")
print("Data will be automatically saved every 5 seconds.")
print("Browse normally in Chrome.")
print("Stop the cell when you finish.")

try:
    while True:

        # Current timestamp
        timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

        # System RAM
        memory = psutil.virtual_memory()

        ram_used_mb = memory.used / (1024 ** 2)
        ram_available_mb = memory.available / (1024 ** 2)

        # Total Chrome RAM
        chrome_ram_mb = 0

        for process in psutil.process_iter(["name", "memory_info"]):
            try:
                if process.info["name"] == "chrome.exe":
                    chrome_ram_mb += (
                        process.info["memory_info"].rss / (1024 ** 2)
                    )

            except (psutil.NoSuchProcess, psutil.AccessDenied):
                pass

        # Create one row
        new_data = pd.DataFrame([{
            "timestamp": timestamp,
            "ram_used_mb": round(ram_used_mb, 2),
            "ram_available_mb": round(ram_available_mb, 2),
            "browser_ram_mb": round(chrome_ram_mb, 2)
        }])

        # Immediately append the data to CSV
        new_data.to_csv(
            file_path,
            mode="a",
            header=False,
            index=False
        )

        print(
            timestamp,
            "| RAM Used:", round(ram_used_mb, 2), "MB",
            "| Available:", round(ram_available_mb, 2), "MB",
            "| Chrome RAM:", round(chrome_ram_mb, 2), "MB"
        )

        time.sleep(INTERVAL)

except KeyboardInterrupt:
    print("\nRAM logging stopped.")
    print("All collected data has already been saved.")
    print("File:", file_path)
```
